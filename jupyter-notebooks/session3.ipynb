## From electrons in a ribbon to the quantum Hall state (***Core***)
We will show how edge states and flat bands appear when including a magnetic field

Here we take a ribbon and we compute its band structure for different values of the magnetic field. The way we include the field is with the Peierls substitution, $t_{ij}\to t_{ij}e^{i\phi_{ij}}$, where the phase $\phi_{ij}=\frac{e}{\hbar}\int_i^j \mathbf A\cdot d\mathbf l$ is set by the flux of $B$ through each plaquette. The colors tell you where each state lives in the ribbon, so what you should look at is which states stay in the bulk, which ones move to the edges as you increase the field, and in which direction the edge states propagate.

```python
from pyqula import geometry
N = 20 # number of unit cells as the width
g = geometry.triangular_ribbon(N) # triangular lattice
Bs = [0.,0.02,0.04] # magnetic field (measured in units of magnetic flux)

# draw a ribbon with the position operator
gs = g.get_supercell(20) 
plt.subplot(1,len(Bs)+1,1) ; plt.title("Position operator")
plt.scatter(gs.r[:,0],gs.r[:,1],c=gs.r[:,1]) ; plt.axis("equal") ; plt.axis("off")

ii = 0
for B in Bs:
    h = g.get_hamiltonian(tij=[-1]) ; h.add_onsite(4.) # generate the Hamiltonian
    h.add_orbital_magnetic_field(B) # add an out-of plane magnetic field
    # plot the bands
    plt.subplot(1,len(Bs)+1,ii+2) ; plt.title("B = "+str(B)) ; ii += 1 # create subplot and increase counter
    (k,e,c) = h.get_bands(kpath=["X","G","X"],operator="yposition") # compute band structure
    plt.scatter(k,e,c=c,vmin=-N/2,vmax=N/2) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) 
    plt.ylim([-4.,4.]) ;  plt.xlim([min(k),max(k)])
    
plt.tight_layout()
```

## The quantum Hall effect in graphene (***Core***)
We will now show the emergence of edge states for a 2d electron gas with a magnetic field

Now we do the same but with an armchair ribbon of the honeycomb lattice. The overall phenomenology is similar, you have bulk states that do not disperse and edge states that propagate, but the underlying electrons are now Dirac electrons instead of quadratic ones, and this changes the Landau levels, which are the flat bands. This is the quantum Hall effect of graphene.

```python
from pyqula import geometry
g = geometry.honeycomb_armchair_ribbon(20) # honeycomb ribbon with armchair edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_orbital_magnetic_field(0.01) # add an out-of plane magnetic field

# draw a ribbon with the position operator
gs = g.get_supercell(20) 
plt.subplot(1,2,1) ; plt.title("Position operator")
plt.scatter(gs.r[:,0],gs.r[:,1],c=gs.r[:,1]) ; plt.axis("equal") ; plt.axis("off")

# plot the bands
plt.subplot(1,2,2) ; plt.title("Band structure")
(k,e,c) = h.get_bands(kpath=["X","G","X"],operator="yposition") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) 
plt.ylim([-1.,1.]) ;  plt.xlim([min(k),max(k)])
```

## Localization of the edge states in the quantum Hall effect (***Challenge***)
Let us now look at where the states are localized as function of energy

Here we take the same ribbon with a magnetic field and we compute the local density of states at several energies, so that we can see directly in real space where the states live. If an energy falls inside a gap between Landau levels, the only states available there are the edge states, whereas if the energy coincides with a Landau level you will mostly see the bulk. It is very useful to compare each panel with the band structure of the previous exercise at the same energy.

```python
from pyqula import geometry
N = 6 # number of plots
ws = np.linspace(0.,0.8,N) # energies

g = geometry.honeycomb_armchair_ribbon(20) # honeycomb ribbon with armchair edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_orbital_magnetic_field(0.01) # add an out-of plane magnetic field

ii = 1 # counter for the subplots
for w in ws:
    plt.subplot(N//3,3,ii); ii += 1 
    (x,y,d) = h.get_ldos(e=w,nrep=20,delta=1e-2,nk=20) # compute local density of states
    plt.title("E = "+str(np.round(w,2))) # set the title
    plt.scatter(x,y,c=d,s=d/np.max(d)*100,vmin=0.,cmap="Blues") ; plt.yticks([]) ; plt.xticks([]) ; plt.axis("equal")# plot
plt.tight_layout() # adjust the subplots
```

## The quantum anomalous Hall effect (***Core***)
Before we considered a system with a magnetic field that has topological edge states. We will now see that topological edge states can appear in the absence of an external magnetic field

The idea is that what we need to have chiral edge states is not a magnetic field, but breaking time reversal symmetry. Here this is done by the internal hoppings of the lattice, with an imaginary second neighbor hopping $t_2 e^{i\phi}$ in the honeycomb lattice, which is what is called the Haldane coupling. The net flux through the unit cell is zero, and yet you get the same chiral edge states and a quantized Hall conductance as in the quantum Hall state, and this is why it is called anomalous. This state is also known as a Chern insulator.

```python
from pyqula import geometry
g = geometry.honeycomb_armchair_ribbon(20) # honeycomb ribbon with armchair edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_haldane(0.1) # add Haldane coupling (second neighbor imaginary hopping)

# draw a ribbon with the position operator
gs = g.get_supercell(20) 
plt.subplot(1,2,1) ; plt.title("Position operator")
plt.scatter(gs.r[:,0],gs.r[:,1],c=gs.r[:,1]) ; plt.axis("equal") ; plt.axis("off")

# plot the bands
plt.subplot(1,2,2) ; plt.title("Band structure")
(k,e,c) = h.get_bands(kpath=["X","G","X"],operator="yposition") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])# plot
```

## Edge states of the quantum anomalous Hall state (***Core***)
Let us now look at the local density of states in a ribbon as function of the energy

As we did for the quantum Hall state, we now take the ribbon with Haldane coupling and we look at the local density of states at different energies. The Haldane coupling opens a topological gap, so the same logic as before applies: inside the gap you can only find edge states, and once you reach the bulk bands the weight spreads across the whole ribbon. The size of the gap is controlled by the Haldane coupling, and this is something you can play with.

```python
from pyqula import geometry
N = 6 # number of plots
ws = np.linspace(0.,0.8,N) # energies

g = geometry.honeycomb_armchair_ribbon(20) # honeycomb ribbon with armchair edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_haldane(0.1) # add Haldane coupling (second neighbor imaginary hopping)

ii = 1 # counter for the subplots
for w in ws:
    plt.subplot(N//3,3,ii); ii += 1 
    (x,y,d) = h.get_ldos(e=w,nrep=20,delta=1e-2,nk=20) # compute local density of states
    plt.title("E = "+str(np.round(w,2))) # set the title
    plt.scatter(x,y,c=d,s=d/np.max(d)*100,vmin=0.,cmap="Blues") ; plt.yticks([]) ; plt.xticks([]) ; plt.axis("equal")# plot
plt.tight_layout() # adjust the subplots
```

## Topological edge states in irregular geometries (***Challenge***)
Before we were considering systems that have translational symmetry in the x direction, showing that topological edge states appear. We will now focus on 0d systems with complex boundaries, showing that those same edge states survive

Here we take a finite island of the honeycomb lattice with Haldane coupling, and we compute the local density of states at different energies. What is important is that the edge states did not appear because of the translational symmetry of the ribbon, they appear because the bulk is topological, and so they should be there for any boundary, even an irregular one. ## Topological states and topological invariants (***Challenge***)
So far we have focused on the emergence of edge states. Let us now address the topological reason for the emergence of edge modes: the Berry curvature

The object that tells us whether a gap is topological or not is the Berry curvature of the occupied band, $\Omega(\mathbf k) = i\left(\langle \partial_{k_x}u_{\mathbf k}|\partial_{k_y}u_{\mathbf k}\rangle - \langle \partial_{k_y}u_{\mathbf k}|\partial_{k_x}u_{\mathbf k}\rangle\right)$, and when we integrate it over the whole Brillouin zone we get an integer, the Chern number
$$C=\frac{1}{2\pi}\int_{BZ} d^2k\,\Omega(\mathbf k)$$
Here we first open a gap with a sublattice imbalance and we plot the Berry curvature in reciprocal space. You will see that it is concentrated around the two valleys, and the question is what happens when you integrate it, and how this changes if the gap is opened by the Haldane coupling instead.

```python
# Now let us do the same, but with a sublattice imbalance
from pyqula import geometry
g = geometry.honeycomb_lattice() # honeycomb lattice
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_sublattice_imbalance(0.5) # add sublattice imbalance (trivial gap)

# let us first compute the band structure, highlighting the Berry curvature
plt.subplot(1,2,1)
(k,e,c) = h.get_bands(operator="Berry") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.]) # plot

# and now lets compute the berry curvature in reciprocal space for the occupied bands
plt.subplot(1,2,2)
(kx,ky,be) = h.get_berry_curvature() # compute band structure
plt.scatter(kx,ky,c=-be/np.max(np.abs(be)),vmin=-1,vmax=1) # plot the Berry curvature in 2D
plt.xlabel("kx") ; plt.ylabel("ky") ; plt.axis("equal") 
plt.xticks([]) ; plt.yticks([])
plt.colorbar(label="Berry curvature",ticks=[]) 
```

## Topological phase transitions in the bulk (***Challenge***)
Let us now see how a topological phase transition takes place in the bulk, when a trivial gap overcomes a topological gap

Now we include both a Haldane coupling and a sublattice imbalance $m$ at the same time, and we compute the Chern number exactly for increasing $m$. The two terms open gaps of a different nature, and when the sublattice imbalance becomes large enough it wins over the Haldane coupling. To go from one regime to the other the gap has to close and reopen, and the Chern number changes at that point. This is a topological phase transition, and it happens in the bulk of the system.

```python
from pyqula import geometry
N = 4 # number of sublattice imbalances
ms = np.linspace(0.,0.5,N) # sublattice imbalances

g = geometry.honeycomb_lattice() # generate a honeycomb lattice


ii = 1 # counter for the subplots
for m in ms:
    plt.subplot(N//2,2,ii); ii += 1 
    h = g.get_hamiltonian() # generate the Hamiltonian
    h.add_haldane(0.05) # add Haldane coupling
    h.add_sublattice_imbalance(m) # add a sublattice imbalance
    (k,e,c) = h.get_bands(operator="Berry") # compute band structure
    plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") 
    plt.xticks([]) ; plt.ylim([-1.,1.]) # plot
    C = np.round(h.get_chern(nk=60)) # compute Chern number
    plt.title("m = "+str(np.round(m,2))+", C="+str(int(C))) # set the title
plt.tight_layout() # adjust the subplots
```

We now take the very same competition between Haldane coupling and sublattice imbalance, but in a ribbon, so that we can follow what happens to the edge states as $m$ increases. Since the edge states are a consequence of the Chern number in the bulk, you should expect that whatever happens to the Chern number is reflected in the edge, and the point here is to check this correspondence yourself by comparing with the previous exercise.

## Finite size effects in topological systems (***Challenge***)
Let us now consider a topological system, and explore how its electronic structure depends on the width of the ribbon. We will generate a Chern insulator by combining exchange coupling and Rashba spin-orbit coupling

Here the Chern insulator is not created with a Haldane coupling but with two couplings together, an exchange field and Rashba spin-orbit coupling, and we compute the band structure of a zigzag ribbon for different widths. Keep in mind that the edge states have a finite localization length, set by the topological gap $E_g$ as $\xi\sim v/E_g$, and the two edges only behave as independent when the ribbon is much wider than $\xi$. When this is not the case, the states of one edge can feel the states of the other edge.

```python
from pyqula import geometry
Ws = [5,10,20] # widths of the ribbons
ii = 1 # counter for the subplots
rashba = 0.1 # strength of Rashba SOC
for W in Ws: # loop over widths
    g = geometry.honeycomb_zigzag_ribbon(W) # honeycomb ribbon with zigzag edge, the number controls the width
    plt.subplot(1,len(Ws),ii); ii += 1 
    h = g.get_hamiltonian() # generate the Hamiltonian
    h.add_exchange([0.,0.,0.8]) # add exchange coupling
    h.add_rashba(rashba) # add Rashba SOC
    (k,e,c) = h.get_bands(kpath=["X","G","X"],operator="yposition") # compute band structure
    plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") 
    plt.xticks([]) ; plt.ylim([-2*rashba,2*rashba]) # plot
    plt.title("Width = "+str(W)) # set the title
plt.tight_layout() # adjust the subplots
```

## The quantum spin Hall state (***Core***)
Let us now focus on a different topological state of matter, the quantum spin Hall effect. This topological state is driven by spin-orbit coupling and features a pair of states in each edge

You can think that a quantum spin Hall insulator is two copies of a Chern insulator, one for spin up and one for spin down, one being the time reversal of the other. We create it by adding intrinsic spin-orbit coupling to the honeycomb lattice, which acts as a Haldane coupling with opposite sign for the two spins, so that $C_\uparrow=-C_\downarrow$. The total Chern number is zero, as it must be with time reversal symmetry, but the spin Chern number $C_s=(C_\uparrow-C_\downarrow)/2$ is not, and this is what gives a pair of edge states in each edge. The colors tell you the location and the spin of each state.

```python
from pyqula import geometry
W = 14 # width of the ribbon
g = geometry.honeycomb_armchair_ribbon(W) # honeycomb ribbon with armchair edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_soc(0.12) # add intrinsic SOC
h.add_onsite(lambda r: 0.6*(np.random.random()-0.5)) # some small disorder to break accidental degeneracies

Ptb = h.get_operator(lambda r: np.sign(r[1])) # projector on the top or bottom part
Sz = h.get_operator("sz") # Sz operator
(ktb,etb,ctb) = h.get_bands(operator=Ptb,kpath=["X","G","X"]) # band projected on the top edge
(ksz,esz,csz) = h.get_bands(operator=Sz,kpath=["X","G","X"]) # band projected in spin

def plot_bands(k,e,c): # function to plot bands
    plt.scatter(k,e,c=c,s=20,cmap="coolwarm") ; plt.xlim([min(k),max(k)]) 
    plt.xticks([]) ; plt.ylim([-1.,1.])

# plot projected on the edge
plt.subplot(1,2,1) ; plt.title("Edge location") ; plot_bands(ktb,etb,ctb)
plt.colorbar(label="Top/bottom",ticks=[],orientation="horizontal")

# plot projected on the spin
plt.subplot(1,2,2) ; plt.title("Spin projection") ; plot_bands(ksz,esz,csz)
plt.colorbar(label="Sz",ticks=[],orientation="horizontal")
```

## The robustness of the quantum spin Hall state (***Core***)
Let us now address the robustness of the quantum spin Hall effect. This topological state features edge states that are robust to perturbations that do not break time reversal symmetry

Here we compute the band structure of the ribbon three times, without disorder, with non-magnetic disorder, and with magnetic disorder. The idea is that the two edge states in each edge are partners under time reversal symmetry, and as long as a perturbation respects that symmetry there is no term that can open a gap between them, the crossing is protected. Magnetic disorder is different, because it breaks time reversal symmetry, and then nothing protects the crossing anymore.

```python
from pyqula import geometry
g = geometry.honeycomb_zigzag_ribbon(10) # honeycomb ribbon with zigzag edge, the number controls the width
h = g.get_hamiltonian() # generate the Hamiltonian
h.add_soc(0.15) # add intrinsic SOC

# first, let us consider a ribbon without disorder
plt.subplot(1,3,1) ; plt.title("No disorder")
(k,e,c) = h.get_bands(operator="bulk") # compute band structure
plt.scatter(k,e,c=c,cmap="Blues") ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])

# now, lets add non-magnetic disorder
plt.subplot(1,3,2) ; plt.title("Non-magnetic disorder")
hd = h.copy() ; hd.add_onsite(lambda r: 1.*(np.random.random()-0.5)) # new Hamiltonian with disorder
(k,e,c) = hd.get_bands(operator="bulk") # compute band structure
plt.scatter(k,e,c=c,cmap="Blues") ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])

# now, lets add magnetic disorder
plt.subplot(1,3,3) ; plt.title("Magnetic disorder")
hd = h.copy() ; hd.add_exchange(lambda r: 1.*(np.random.random(3)-0.5)) # new Hamiltonian with disorder
(k,e,c) = hd.get_bands(operator="bulk") # compute band structure
plt.scatter(k,e,c=c,cmap="Blues") ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])

plt.tight_layout() # adjust plots
```

## Topological invariant of the quantum spin Hall state (***Challenge***)
Let us now see how the topological invariant of a gapped system with time reversal symmetry allows to predict the existence of edge states

For a system with time reversal symmetry the Chern number is always zero, so we need a different invariant, and this is the $\mathbb Z_2$ invariant $\nu\in\{0,1\}$. Here we open a topological gap with spin-orbit coupling and a trivial gap with a sublattice imbalance, we compute $\nu$ in the bulk for increasing imbalance, and at the same time we compute the band structure of a ribbon. A value $\nu=1$ means an odd number of pairs of edge states that no perturbation respecting time reversal symmetry can remove, whereas for $\nu=0$ they can be gapped out, and you can follow this correspondence between bulk and edge in the plots.

```python
from pyqula import geometry, ribbon, topology
ms = np.linspace(0.1,0.8,3) # sublattice imbalances

ii = 1 # counter for the subplots
for m in ms:
    # bulk electronic structure and topological invariant
    plt.subplot(2,len(ms),ii) # new subplot 
    g = geometry.honeycomb_lattice() # honeycomb lattice
    h = g.get_hamiltonian(tij=[1.,0.]) ; h.set_filling(0.5) # generate the Hamiltonian at half filling
    h.add_soc(0.1) # add SOC that drives QSH
    h.add_sublattice_imbalance(m) # add a sublattice imbalance
    (k,e) = h.get_bands() # compute band structure
    plt.scatter(k,e,c="black") ; plt.xlabel("Momentum") ; plt.ylabel("Energy") 
    plt.xticks([]) ; plt.ylim([-1.,1.]) # plot
    z2 = topology.z2_invariant(h) # compute the Z2 invariant
    plt.title("m = "+str(np.round(m,2))+", $Z_2 =$"+str(z2)) # set the title
    # now electronic structure of a ribbon
    plt.subplot(2,len(ms),ii+len(ms)) # new subplot 
    h = ribbon.bulk2ribbon(h,n=20) # create a ribbon
    (k,e,c) = h.get_bands(operator="yposition") # compute band structure
    plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])
    ii += 1
plt.tight_layout() # adjust the subplots
```

## The valley Hall effect (***Challenge***)
Let us now address how an electrically controllable topological edge state can appear in graphene multilayers

The valley Hall effect is not protected by a fundamental symmetry, but by an approximate one, the valley symmetry. A bilayer of graphene with an interlayer bias opens a gap whose Berry curvature has opposite sign in the two valleys, $\Omega_{K'}(\mathbf k)=-\Omega_K(\mathbf k)$, so the total Chern number is zero but each valley carries its own Chern number $C_K=-C_{K'}$. As long as there is no scattering between the valleys, each Dirac cone behaves as an independent topological system, and at a domain wall between regions with opposite bias you get edge states, that can be switched electrically by reversing the bias.

```python
from pyqula import specialhamiltonian # use 
h = specialhamiltonian.multilayer_graphene(l=[0,1],ti=0.3) # Hamiltonian of an AB bilayer 

# first without an interlayer bias
plt.subplot(1,2,1) ; plt.title("No interlayer bias")
(k,e) = h.get_bands() # compute band structure
plt.scatter(k,e,c="black") ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.])


# now with an interlayer bias
plt.subplot(1,2,2) ; plt.title("Finite interlayer bias")
h.add_onsite(lambda r: 0.1*np.sign(r[2])) # bias difference between layers
(k,e,c) = h.get_bands(operator="berry") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-1.,1.]) # plot
plt.colorbar(label="Berry curvature",ticks=[],orientation="horizontal")
```

```python
# let us now compute a ribbon of bilayer AB graphene
from pyqula import specialhamiltonian # use 
h = specialhamiltonian.multilayer_graphene(l=[0,1],ti=0.3) # Hamiltonian of an AB bilayer 
from pyqula import ribbon
h = ribbon.bulk2ribbon(h,n=20) # generate a ribbon from the bulk Hamiltonian
h.add_onsite(lambda r: 0.2*np.sign(r[2])) # bias difference between layers

# compute the expectation value of the location
plt.subplot(1,2,1) ; plt.title("Location")
(k,e,c) = h.get_bands(operator="yposition") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-.5,.5])# plot

# compute the expectation value of the valley
plt.subplot(1,2,2) ; plt.title("Valley")
(k,e,c) = h.get_bands(operator="valley") # compute band structure
plt.scatter(k,e,c=c) ; plt.xlabel("Momentum") ; plt.ylabel("Energy") ; plt.xticks([]) ; plt.ylim([-.5,.5])# plot

```
